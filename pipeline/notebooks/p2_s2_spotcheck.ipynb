```python
import os, sys
sys.path.insert(0, os.path.abspath('..'))
from IPython.display import Image, display
from src import ee_client, params, s2, sites
ee_client.init()
p = params.load()
SITE, START, END = 'tismana_1', '2026-08-01', '2026-11-01'
aoi = sites.aoi(sites.get(SITE))
df = s2.to_frame(s2.scene_rows(aoi, START, END, p).getInfo(), p['s2']['min_clear_frac'])
print(len(df), 'scenes,', int(df.obs.sum()), 'clear obs')
df
```
